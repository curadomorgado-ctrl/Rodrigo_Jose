```python
from matplotlib.axis import XAxis
import numpy as np
import cv2 as cv
import plotly.express as px
import glob
from matplotlib.pyplot import scatter, xlabel, ylabel
from skimage import io

#Paths to the images
apple_path='Datasets/FRUITS/APPLES';
orange_path='Datasets/FRUITS/ORANGES';
peache_path='Datasets/FRUITS/PEACHES';

# Procurar todas as imagens .JPG dentro de cada pasta # glob.glob() devolve uma lista com os caminhos das imagens
#Get file names from directories
ap_imgs=glob.glob(apple_path+'/*.JPG'); ##serve o mesmo efeito que o list dir
or_imgs=glob.glob(orange_path+'/*.JPG');
pe_imgs=glob.glob(peache_path+'/*.JPG');

#Compute number of images with apples, oranges and peaches
n_ap=len(ap_imgs);
n_or=len(or_imgs);
n_pe=len(pe_imgs);
n_fr=n_ap+n_or+n_pe;
print(f"Existem {n_ap} maçãs")
print(f"Existem {n_or} laranjas")
print(f"Existem {n_pe} pêssegos")
print(f"Existem {n_fr} frutas no total")

#Empty feature vectors
cr=np.zeros(n_fr);#Color ratio
sr=np.zeros(n_fr);#Shape ratio
labels=[];

imgs=[ap_imgs,or_imgs,pe_imgs];
fnames=['Apples','Oranges','Peaches']
k=0;

for i in ap_imgs:
    image_total=io.imread(i)
    vermelho=image_total[150:200,150:200,0]
    verde=image_total[150:200,150:200,1]

    # Converter a imagem para grayscale # 
    # # A imagem deixa de ter 3 canais (RGB/BGR) # e passa a ter apenas um valor de intensidade # por pixel.
    Image_grey_scale=cv.cvtColor(image_total,cv.COLOR_BGR2GRAY)
    Image_equalized=cv.equalizeHist(Image_grey_scale)
    # Transformar a imagem grayscale numa imagem binária. #
    #  # Se o pixel tiver valor <= 100: # passa para 0 (preto) # 
    # # Se o pixel tiver valor > 100: # passa para 255 (branco) #
    #  # _ -> valor do threshold que não precisamos de guardar # Imagem_final -> imagem binária resultante
    _, Imagem_final=cv.threshold(Image_equalized,100,255,cv.THRESH_BINARY)#estamos a passar uma imagem grayscale (0 e 1)

    #Find the indexes of the contour, vê onde é que há mudanças de cor
    [row,col] = np.where(np.diff(Imagem_final[0:-5,0:-5])!=0);#so vai ficar com os primeiros pontos 

    #Compute the shape ratio
    sr[k]=(max(col)-min(col))/(max(row)-min(row));

    valor_medio_vermelho= np.mean(vermelho)
    valor_medio_verde=np.mean(verde)

    ratio_apples_colour=valor_medio_vermelho/valor_medio_verde
    cr[k]=ratio_apples_colour     
    labels.append('Apples')
    k=k+1

for i in or_imgs:
    image_total=io.imread(i)
    vermelho=image_total[150:200,150:200,0]
    verde=image_total[150:200,150:200,1]
    Image_grey_scale=cv.cvtColor(image_total,cv.COLOR_BGR2GRAY)
    Image_equalized=cv.equalizeHist(Image_grey_scale)
    _, Imagem_final=cv.threshold(Image_equalized,100,255,cv.THRESH_BINARY)#estamos a passar uma imagem grayscale (0 e 1)
    #Find the indexes of the contour
    [row,col] = np.where(np.diff(Imagem_final[0:-5,0:-5])!=0);#so vai ficar com os primeiros pontos 
    #Compute the shape ratio
    sr[k]=(max(col)-min(col))/(max(row)-min(row));
    valor_medio_vermelho= np.mean(vermelho)
    valor_medio_verde=np.mean(verde)
    ratio_oranges_colour=valor_medio_vermelho/valor_medio_verde
    cr[k]=ratio_oranges_colour
    labels.append('Oranges')
    k=k+1

for i in pe_imgs:
    image_total=io.imread(i)
    vermelho=image_total[150:200,150:200,0]
    verde=image_total[150:200,150:200,1]
    Image_grey_scale=cv.cvtColor(image_total,cv.COLOR_BGR2GRAY)
    Image_equalized=cv.equalizeHist(Image_grey_scale)
    _, Imagem_final=cv.threshold(Image_equalized,100,255,cv.THRESH_BINARY)#estamos a passar uma imagem grayscale (0 e 1)
    #Find the indexes of the contour
    [row,col] = np.where(np.diff(Imagem_final[0:-5,0:-5])!=0);#so vai ficar com os primeiros pontos 
    #Compute the shape ratio
    sr[k]=(max(col)-min(col))/(max(row)-min(row));
    valor_medio_vermelho= np.mean(vermelho)
    valor_medio_verde=np.mean(verde)
    ratio_peaches_colour=valor_medio_vermelho/valor_medio_verde
    cr[k]=ratio_peaches_colour
    labels.append('Peaches')
    k=k+1

px.scatter(x=cr,y=sr,color=labels,labels=dict(x="Color Ratio", y="Shape Ratio",color="Fruit"))
```
