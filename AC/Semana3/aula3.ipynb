BINARY EVALUATION METRICS
Valores previstos

```python
##Binary evaluation metrics
import numpy as np
from sklearn.metrics import confusion_matrix
def decision_function(f1, f2):
    previsao = -f1 + f2
    if previsao >= 0:
        return 1
    else:
        return -1

X = [(0,1), (1, 2), (2, 4), (3, 5),(1, 0),(2, 1),(3, 2),(4, 0)]
y_predictor=[]
for f1, f2 in X:
    y_predictor.append(decision_function(f1, f2))
    print(f1, f2, "Valor previsto é", decision_function(f1, f2))
```

Matriz confusão

```python
##Matriz confusão

false_positive = false_negative = true_positive = true_negative = 0

y = [1, 1, 1, -1, -1, -1, -1, 1]

for i in range(len(y)):
    real = y[i]
    prev = y_predictor[i]
    if real == 1 and prev == 1:
        true_positive += 1
    elif real == 1 and prev == -1:
        false_negative += 1
    elif real == -1 and prev == 1:
        false_positive += 1
    else:
        true_negative += 1

confusao = [[true_positive, false_negative],
            [false_positive, true_negative]]
print(confusao)
```

Calcular as métricas

```python
#Calcular as métricas
accuracy = (true_positive + true_negative) / (true_positive + true_negative + false_negative + false_positive)
print(f"Accuracy:    {accuracy:.2%}")

sensitivity = true_positive / (true_positive + false_negative)
print(f"Sensitivity: {sensitivity:.2%}")

specificity = true_negative / (true_negative + false_positive)
print(f"Specificity: {specificity:.2%}")

precision = true_positive / (true_positive + false_positive)
print(f"Precision:   {precision:.2%}")

recall = true_positive / (true_positive + false_negative)
print(f"Recall:      {recall:.2%}")

f1_score = 2 * precision * recall / (precision + recall)
print(f"F1-score:    {f1_score:.2%}")
```

Multiclass evaluation metrics

```python
##Binary evaluation metrics
def decision_function_multi_class(f1, f2):
    if f1 >= 0 and f2 >= 0:
        return 1
    if f1 >= 0 and f2 < 0:
        return 2
    if f1 < 0 and f2 < 0:
        return 3
    if f1 < 0 and f2 >= 0:
        return 4

X_multi = [(0, 1), (1, 2), (2, 4), (3, -5), (1, -1), (2, -1),
     (-3, -2), (-4, -1), (-1, -2), (-3, 4), (-1, 1), (-2, 2)]

y_predictor_multi=[]
for f1, f2 in X_multi:
    y_predictor_multi.append(decision_function_multi_class(f1, f2))
    print(f1, f2, "Valor previsto é", decision_function_multi_class(f1, f2))
```

Matriz confusão multiclasse

```python
##Matriz confusão
y_multiclass_true = [1, 1, 2, 2, 2, 2, 2, 3, 3, 4, 1, 2]

true_1 = true_2 = true_3 = true_4 = 0
false1_2 = false1_3 = false1_4 = 0
false2_1 = false2_3 = false2_4 = 0
false3_1 = false3_2 = false3_4 = 0
false4_1 = false4_3 = false4_2 = 0

for i in range(len(y_multiclass_true)):
    real_multi = y_multiclass_true[i]
    prev_multi = y_predictor_multi[i]
    if real_multi == 1 and prev_multi == 1:
        true_1 += 1
    elif real_multi == 2 and prev_multi == 2:
        true_2 += 1
    elif real_multi == 3 and prev_multi == 3:
        true_3 += 1
    elif real_multi == 4 and prev_multi == 4:
        true_4 += 1
    elif real_multi == 2 and prev_multi == 1:
        false1_2 += 1
    elif real_multi == 3 and prev_multi == 1:
        false1_3 += 1
    elif real_multi == 4 and prev_multi == 1:
        false1_4 += 1
    elif real_multi == 1 and prev_multi == 2:
        false2_1 += 1
    elif real_multi == 3 and prev_multi == 2:
        false2_3 += 1
    elif real_multi == 4 and prev_multi == 2:
        false2_4 += 1
    elif real_multi == 1 and prev_multi == 3:
        false3_1 += 1
    elif real_multi == 2 and prev_multi == 3:
        false3_2 += 1
    elif real_multi == 4 and prev_multi == 3:
        false3_4 += 1
    elif real_multi == 1 and prev_multi == 4:
        false4_1 += 1
    elif real_multi == 3 and prev_multi == 4:
        false4_3 += 1
    elif real_multi == 2 and prev_multi == 4:
        false4_2 += 1

confusao_multi = [[true_1,   false2_1, false3_1, false4_1],
                  [false1_2, true_2,   false3_2, false4_2],
                  [false1_3, false2_3, true_3,   false4_3],
                  [false1_4, false2_4, false3_4, true_4]]
for linha in confusao_multi:
    print(linha)
```

Calcular as métricas

```python
accuracy_multi = (true_1 + true_2+ true_3 + true_4) / (true_1 + true_2+ true_3 + true_4 + false1_2+false1_3+false1_4+false2_1+false2_3+false2_4+false3_1+false3_2+false3_4+false4_1+false4_2+false4_3  )
print(f"Accuracy:    {accuracy_multi:.2%}")

##sensitivity for line of true ones
sensitivity_multi_1=true_1/(true_1+false1_2+false1_3+false1_4);
sensitivity_multi_2=true_2/(true_2+false2_1+false2_3+false2_4);
sensitivity_multi_3=true_3/(true_3+false3_1+false3_2+false3_4);
sensitivity_multi_4=true_4/(true_4+false4_1+false4_2+false4_3);

K=4; #numero de classes
macro_sensitivity=(sensitivity_multi_1+sensitivity_multi_2+sensitivity_multi_3+sensitivity_multi_4)/K;
print(f"Macro Sensitivity: {macro_sensitivity:.2%}")

##precision for the colum of predicted ones
precision_multi_1=true_1/(true_1+false2_1+false3_1+false4_1);
precision_multi_2=true_2/(true_2+false1_2+false3_2+false4_2);
precision_multi_3=true_3/(true_3+false1_3+false2_3+false4_3);
precision_multi_4=true_4/(true_4+false2_4+false3_4+false1_4);

macro_precision=(precision_multi_1+precision_multi_2+precision_multi_3+precision_multi_4)/K;
print(f"Macro Precision: {macro_precision:.2%}")

macro_recall=(sensitivity_multi_1+sensitivity_multi_2+sensitivity_multi_3+sensitivity_multi_4)/K;
print(f"Recall: {macro_recall:.2%}")

macro_f1_score=(2*macro_precision*macro_sensitivity)/(macro_precision+macro_sensitivity)
print(f"Macro F-1 Score: {macro_f1_score:.2%}")
```
