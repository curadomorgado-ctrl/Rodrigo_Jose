# Importar os dados (Aula 5)

```python
import scipy.io as sio
import numpy as np
import pandas as pd

# Caminho da pasta
path = r"C:\Users\josem\OneDrive\Ambiente de Trabalho\Zé Miguel\UNIVERSIDADE\5ANO\Rodrigo_Jose\AC\Semana5"

# Carregar o dataset 1
data = sio.loadmat(path + r"\dataset2.mat") #mudar para 2 se quiser
# Extrair os dados
X1 = data['data_2d'][0][0][0]
y1 = data['data_2d'][0][0][1][0]
```

# Exercise 1

```python
# Normalização Z-score
X1_normed = (X1.T - np.mean(X1.T, axis=0)) / np.std(X1.T, axis=0)

C = np.cov(X1_normed.T)
print("C=\n" + str(C) + "\n")

# Compute eigenvalues and eigenvectors
eigVal, eigVec = np.linalg.eigh(C)

# Order eigenvalues in descending order
ind = np.argsort(eigVal)[::-1]
eigVal = eigVal[ind]
eigVec = eigVec[:, ind]

# Eigenvectors
print("Eigenvectors=\n" + str(eigVec))

# Eigenvalues
print("Eigenvalues=\n" + str(eigVal) + "\n")

# Two most important eigenvectors
print("Two most important Eigenvectors=\n" + str(eigVec[:, 0:2]) + "\n")

# Compute variance explained by the  most important PC
explained_variance = (eigVal[0]) / np.sum(eigVal) * 100

print("Variance explained by the two most important PCs:")
print("R(%) =", explained_variance)

# Projeção dos dados no primeiro componente principal (PC1)
Xp = X1_normed @ eigVec[:, 0]          # shape (400,)
import plotly.express as px
px.scatter(x=Xp, y=np.zeros(len(Xp)), color=y1.astype(str),
           title="Data after PCA",
           labels=dict(x="PC1", y="", color="Class")).show()
```

# Exercise 2

```python
# Ler o ficheiro: separador ';', decimal ',' e saltar a linha vazia (linha 2)
df = pd.read_csv(
    r"C:/Users/josem/OneDrive/Ambiente de Trabalho/Zé Miguel/UNIVERSIDADE/5ANO/Rodrigo_Jose/AC/datasets/CORK_STOPPERS.csv",
    sep=";",
    decimal=",",
    skiprows=[1]
)

# Definir as características a analisar
features = ["ART", "N.1", "PRT", "ARM", "PRM",
            "ARTG", "NG", "PRTG", "RAAR", "RAN"]

# Definir as categorias de qualidade
quality = ["Super", "Average", "Poor"]

# Converter os códigos numéricos da coluna C em categorias
df["C"] = df["C"].map({
    1: quality[0],
    2: quality[1],
    3: quality[2]
})
# Extrair as características como array NumPy (n_amostras x 10)
X_cork = df[features].values

# Normalize
# Normalização Z-score
X_cork = (X_cork - np.mean(X_cork, axis=0)) / np.std(X_cork, axis=0)

print(X_cork)
```

# EX2.1

```python
from sklearn.decomposition import PCA

pca = PCA()
pca.fit(X_cork)

print('Eigen values / Explained variance: ')
print(pca.explained_variance_)

print('\nSum of eigen values: ' + str(np.sum(pca.explained_variance_)))

# Eigen vectors / principal components
print('\nEigen vectors / Principal components: ')
W = pca.components_.T
print(W)
```

# Ex2.2  (SLIDE 11 DA AULA 4)

```python
# Scree plot: gráfico dos valores próprios (explained variance) de cada componente principal

# Eixo x: números das componentes (PC1, PC2, ...), de 1 até ao número total de componentes
# Eixo y: valor próprio de cada componente, ou seja, a variância dos dados nessa direção
fig = px.scatter(x=np.arange(1, len(pca.explained_variance_) + 1),
                 y=pca.explained_variance_,
                 labels=dict(x="PC", y="Explained Variance"),
                 title="Scree plot")

# Linha horizontal em y = 1 (critério de Kaiser):
# retêm-se as componentes com valor próprio acima de 1
fig.add_hline(y=1, line_width=3, line_dash="dash", line_color="red")

# Aumenta o tamanho dos marcadores para facilitar a leitura do gráfico
fig.update_traces(marker_size=10)

# Mostra o gráfico
fig.show()
```

# EX2.3

```python
# Criar o PCA com apenas 1 componente (critério de Kaiser: só a PC1 tem valor próprio > 1)
pca2 = PCA(n_components=1)

# Ajustar o PCA aos dados normalizados e projetá-los na PC1
# X1D fica com shape (150, 1): uma coordenada por amostra
X1D = pca2.fit_transform(X_cork)
print(np.shape(X1D))   # (150, 1)

# Gráfico dos dados projetados na PC1
# Eixo x: valor de cada amostra na PC1
# Eixo y: fixo em 0, porque a projeção é unidimensional
# Cor: classe de cada amostra (Super, Average, Poor)
fig = px.scatter(x=X1D[:, 0], y=np.zeros(X1D.shape[0]),
                 color=df["C"],
                 labels=dict(x="PC1", y="", color="Cork Stopper"),
                 title="Dados projetados na PC1 (Kaiser)")

# Aumenta o tamanho dos marcadores para facilitar a leitura
fig.update_traces(marker_size=8)

# Mostra o gráfico
fig.show()
```
