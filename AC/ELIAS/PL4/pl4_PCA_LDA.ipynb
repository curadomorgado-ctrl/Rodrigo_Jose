Consider the following data, where lines represent samples and columns features:

1. Normalize/standardize your data features such that each feature will have
zero mean and standard deviation of 1.
2. Compute PCA by implementing explicitly the PCA algorithm and:
• Identify the eigenvectors and eigenvalues;
• identify the two most important eigenvectors and what is the % of
variance explained by them.


```python
import pandas as pd
import numpy as np

X = pd.DataFrame({"Class1": [7, 4, 6, 8, 8, 7, 5, 9, 7, 8],
                 "Class2": [4, 1, 3, 6, 5, 2, 3, 5, 4, 2],
                 "Class3": [3, 8, 5, 1, 7, 9, 3, 8, 5, 2]})
X
m = np.mean(X, axis=0) # mean of each colum
sd = np.std(X, axis=0) # sd of each col
norm_X = (X - m) / sd # standard normalization

matrix = np.cov(norm_X, rowvar=False)
print(matrix)
# w: valores proprios; v:vetores proprios
w, v = np.linalg.eig(matrix)

ind = np.argsort(w)
w_s = w[np.flip(ind)]
v_s = v[np.flip(ind)]

# print(w, ind)
print("Eigenvectores\n", v_s)
print("\nEigenvalues: ", w_s)


R1 = w_s[0]**2 / sum(np.square(w))
R2 = w_s[1]**2 / sum(np.square(w))
print("Percentage of variance of the two most important eigvalues: ",R1 + R2)
```

Inspect the utility of projecting the datasets represented in the figure to one
dimension by using PCA. Explain graphically your assumptions. (Implement
explicitly the PCA algorithm!)
For your convenience, these exemplary datasets are available at UCStudent
(dataset1.mat and dataset2.mat).

```python
import scipy.io
import plotly.express as px

folder = "D:/UC.library/Mestrado/1_ANO/2_semestre/RP/PL/PL4/data/"
df1 = scipy.io.loadmat(folder + "dataset1.mat")
df2 = scipy.io.loadmat(folder + "dataset2.mat")


### Dataset for df1 #####
x1 = df1["data_2d"][0][0][0]
y1 = df1["data_2d"][0][0][1][0]


px.scatter(x1, x=x1[0], y=x1[1], color=y1).show()

# Standard normalisation
for i in range(0,2):
    mean_1 = np.mean(x1[i])
    sd_1 = np.std(x1[i])
    # print(mean_1,' ',sd_1)
    x1[i] = (x1[i] - mean_1) / sd_1
px.scatter(x1, x=x1[0], y=x1[1], color=y1, title="Data normalised").show() 


covar_matrix = np.cov(x1, rowvar=True)
print("Covariance matrix:\n",covar_matrix)
eig_val, eig_vec = np.linalg.eig(covar_matrix)

index = np.argsort(eig_val)
eig_val = eig_val[np.flip(index)]
eig_vec = eig_vec[np.flip(index)]

print('Eigenvectoers:\n',eig_vec, '\nEigenvalues: ', eig_val)

Xp = np.matmul(eig_vec[:,0], x1)

px.scatter(x=Xp, y=np.zeros(np.shape(x1)[1]), color=y1, title="Data after PCA", labels=dict(x="PC1", y = '', color = 'Class')).show() 
```

```python
import pandas as pd
import numpy as np
import plotly.express as px

cork = pd.read_excel("D:/UC_library/Mestrado/1_ANO/2_semestre/RP/PL/data_sets/CORK_STOPPERS.XLS",
                     sheet_name="Data",
                     usecols=['C', 'ART', 'N.1', 'PRT', 'ARM', 'PRM', 'ARTG', 'NG', 'PRTG', 'RAAR', 'RAN'])

cork = cork.dropna()


classes = ['Super', 'Average', 'Poor']

cork['C'] = cork['C'].map({1 : classes[0], 2 : classes[1], 3 : classes[2]})

X = cork.to_numpy()[:,1:]
Y = cork.to_numpy()[:,0]
X = X.astype(float)

# Normalize
X = (X - np.mean(X, axis=0))/np.std(X, axis=0)

print(X)
```

### 1. Find the directions of the Principal Components and the variance explained by each one. (Use PCA from sklearn.decomposition !)

```python
from sklearn.decomposition import PCA

pca = PCA()
pca.fit(X)

print('Eigen values / Explained variance: ')
print(pca.explained_variance_)

print('\nSum of eigen values: ' + str(np.sum(pca.explained_variance_)))

# Eigen vectors / principal components
print('\nEigen vectors / Principal components: ')
W = pca.components_.T
print(W)
```

### 2. Plot the eigenvalues. Use Kaiser test and the Scree test to decide how many components to retain. What is the variance retained in each case?

```python
import plotly.express as px

fig = px.scatter(x = np.arange(1,11,1), y = pca.explained_variance_,
                 labels=dict(x = 'PC', y = 'Explained variance')
                 )
fig.add_hline(y = 1, line_width = 3, line_dash = 'dash', line_color = 'red')
fig.update_traces(marker_size = 10)
fig.show()
```

According to Kaiser test 1 PC is enough - number of eigenvalue above 1 when data is normalized

According to Scree test we can use 5 initial PCs - subjectively from PC 6 thstrs not much variation

```python
print('Variance (%) retainde according kaiser: ' + str(pca.explained_variance_[0]**2/(np.sum(pca.explained_variance_**2)) * 100))
print('Variance (%) retainde according Scree: ' + str(np.sum(pca.explained_variance_[0:6]**2)/(np.sum(pca.explained_variance_**2)) * 100))
```
