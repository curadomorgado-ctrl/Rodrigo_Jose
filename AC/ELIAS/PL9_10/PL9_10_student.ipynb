## Support Vector Machines (Hard-Margin)
#### Consider the following training data, where each column of X is a feature and y are the labels associated to each sample:

Compute w.

```python
import numpy as np
import pandas as pd
from GivenCode import clipAlphasJ, selectJrandom, simplifiedSMO

X = np.array([(-2, -1, 0, 1, 2, 0, 3, 4), (-1, -1, 0, 0, 0, 2, 2, 4)]).T
y = np.array([1,1,1,1,-1,-1,-1,-1])


alphas = simplifiedSMO(X, y)
# print(alphas)

w = np.zeros((1,2))
for i in range(y.shape[0]):
    w += (alphas[i]*y[i]*X[i,0])
w = np.asmatrix(w).T
print(w)




```

#### Compute the separation margin width.

Margin width is given by $\frac{2}{||w||}$ ,  $||w|| = \sqrt{w_1^2 + w_2^2}$

```python
print('Margin width: ' + str(2/np.linalg.norm(w)))
```

## Support Vector Machines (Soft-Margin & Non-Linear)
1. Apply and analyse SVM applied to the Cork Stoppers Data Set in order
to discriminate between classes ω1 (Super), ω2 (Average) and by considering the features N (Total number of defects) and PRT/10 (Total perimeter
of the defects (in pixels) normalized by 10)

```python
import pandas as pd
import numpy as np

cork = pd.read_excel('D:/UC_library/Mestrado/1_ANO/2_semestre/RP/PL/data_sets/CORK_STOPPERS.XLS', sheet_name='Data',
                     usecols=['C', 'N.1', 'PRT'])
cork = cork.dropna()
cork = cork[cork["C"].isin([1, 2])]
cork['PRT'] = cork['PRT']/10

df = np.array(cork)
# print(df[:, 0:3])
```

• Define the training and test datasets (consider 50% of the patterns
for each one)

• Consider a linear kernel by considering different C values (e.g., [2^−20,...,2^20]).
Determine the value of C that lead to the reduced error in testing.
To decide about this you should run at least 10 times the algorithm
involving different random selections of the training and testing data.

```python
import sklearn.svm as svm
from sklearn.model_selection import train_test_split

runs = np.arange(0.0,10.0,1.0)
C_exp = np.arange(-20.0,20.0,2.0)

err = np.zeros((len(C_exp), len(runs)))

for i in runs:
    x_train, x_test, y_train, y_test = train_test_split(df[:,1:3], df[:,0], test_size=0.5, random_state=int(i))

    x_train_norm = (x_train - np.mean(x_train, axis=0) )/ np.std(x_train, axis=0)
    x_test_norm = (x_test - np.mean(x_train, axis=0) )/ np.std(x_train, axis=0)

    for idx_c,c in enumerate(C_exp):
        C_value = 2 ** c

        # 1. Create model
        clf = svm.SVC(C=C_value, kernel='linear')

        # 2. Train model
        clf.fit(x_train_norm, y_train)

        # 3. Predict
        y_pred = clf.predict(x_test_norm)
       
        # 4. Accuracy
        acc = np.mean(y_pred == y_test)

        # Error 
        err[idx_c - 1,  int(i) - 1] = 1 - acc


print(err)


```
