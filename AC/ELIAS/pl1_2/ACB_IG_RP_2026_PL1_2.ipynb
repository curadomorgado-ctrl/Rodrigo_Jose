## 1.1 Excel Data

### 1-Open the file CORK_STOPPERS.XLS in Excel, which is available at {UCStudent-Folder Data}. The file contains measurements performed automatically by an image processing system on 150 cork stoppers belonging to three classes. Each image was digitized with an adequate threshold in order to enhance the defects. The first step is to read the available data description in the Description separator. The data description tells us that the dataset is composed by a set of 10 discriminant features. Looking at the Data separator we can observe that the columns C to L correspond each one to one feature. The A column enumerates the patterns and the B column defines the class to which a given pattern belongs. Three categories are considered for the Cork Stoppers: 1-Quality Super; 2- Quality Average; 3- Quality Poor. 

```python
import numpy as np
import plotly.express as px
import pandas as pd
import os

datasets_folder = os.path.abspath(os.path.join(os.getcwd() ,"../..")) + "/Datasets/" #Set the path to the datasets folder, which is assumed to be two levels up from the current working directory and then inside a "Datasets" folder. Adjust this path as necessary based on your actual directory structure.

#Read XLS file and create DataFrame, selecting only relevant columns and dropping rows with NaN values.    
xlsFile=pd.read_excel(datasets_folder + "CORK_STOPPERS.XLS",sheet_name="Data",
                      usecols=["C","ART","N.1","PRT","ARM","PRM","ARTG","NG","PRTG","RAAR","RAN"])
CorkStopp=pd.DataFrame(data=xlsFile.dropna()) #Remove NaN because of initial empty lines

Classes=["Super","Average","Poor"]

# Map numerical values to strings for better readability, assuming 1, 2, 3 correspond to "Super", "Average", "Poor" respectively.
CorkStopp['C'] = CorkStopp['C'].map({1: Classes[0], 2: Classes[1], 3: Classes[2]})
```

### 2-Make a scatter-plot that represents the Cork Stoppers in a space generated by  N  and ARM? Can you visualize the three categories/classes?

```python
# Create a scatter plot using Plotly Express, with 'N.1' on the x-axis, 'ARM' on the y-axis, and color/symbol based on the 'C' column (Cork Stoppers)
fig = px.scatter(CorkStopp, x = 'N.1', y = 'ARM',
                color = 'C', symbol = 'C',
                labels = {'N.1':'N','ARM':'ARM','C':'Cork Stoppers'})
fig.update_traces(marker_size=10)
fig.show()
```

#### 3-Make a 3D scatter plot representing the Cork Stoppers in a data space spanned by features N, ARM, and PRM.

```python
# Create a 3D scatter plot using Plotly Express, with 'N.1' on the x-axis, 'ARM' on the y-axis, 'PRM' on the z-axis, and color based on the 'C' column (Cork Stoppers)
fig = px.scatter_3d(data_frame = CorkStopp, x = 'N.1', y = 'ARM', z = 'PRM',
                    color = 'C',
                    labels={'N.1':'N','ARM':'ARM','PRM':'PRM','C':'Cork Stoppers'},
                    height=600,width=800)
fig.update_traces(marker_size=10)
fig.show()
```

# 2 - Feature extraction - Fruits Dataset


Consider an automatic system, represented in Fig. 1 that aims to separate three types of fruits (apples, peaches or oranges) in different containers. The main system component is a PC equipped with a camera that takes pictures from the fruits in a conveyor belt; extract features from the fruit images; and classifies them in three different classes. In this class we will face the problem of extracting features from fruit pictures. Consider the Fruits dataset available at UCStudent as the pictures taken by the PC camera.

The dataset contains color pictures of 22 apples, 15 oranges and 36 peaches. Each pixel  in the pictures is codified in 256 levels RGB. 

1. Inspect about possible features.
2. Develop code to extract the features.
3. Plot the features and inspect about how good they are in discriminating
the different fruit types.
4. Answer to the following equations:
    * What is the dimensionality of the problem?
    * How many examples are available in the dataset?

```python
from matplotlib.pyplot import gray
import numpy as np
import cv2 as cv
import plotly.express as px
import glob
from skimage import io
import os

datasets_folder = os.path.abspath(os.path.join(os.getcwd() ,"../..")) + "/Datasets/" #Set the path to the datasets folder, which is assumed to be two levels up from the current working directory and then inside a "Datasets" folder. Adjust this path as necessary based on your actual directory structure.

#Paths to the images
apple_path = datasets_folder + 'FRUITS/APPLES'
orange_path = datasets_folder + 'FRUITS/ORANGES'
peache_path = datasets_folder + 'FRUITS/PEACHES'


#Get file names from directories 
ap_imgs_file_names = glob.glob(apple_path+'/*.JPG')
or_imgs_file_names = glob.glob(orange_path+'/*.JPG')
pe_imgs_file_names = glob.glob(peache_path+'/*.JPG')

#Compute number of images with apples, oranges and peaches
n_ap = len(ap_imgs_file_names)
n_or = len(or_imgs_file_names)
n_pe = len(pe_imgs_file_names)
n_fr = n_ap + n_or + n_pe

#Create empty feature vectors to store the computed features for all the images based on the total number of images (n_fr)
cr = np.zeros(n_fr);#Color ratio
sr = np.zeros(n_fr);#Shape ratio
m_h = np.zeros(n_fr);#Mean Hue
m_s = np.zeros(n_fr);#Mean Saturation
m_v = np.zeros(n_fr);#Mean Value
contrast = np.zeros(n_fr);#Contrast
labels = [] #List to store the labels of the fruit class for each image, which will be used for plotting the fruits in the feature space with different colors based on their class

imgs_filenames = [ap_imgs_file_names,or_imgs_file_names,pe_imgs_file_names] #List of lists to store the file names of the images for each fruit class, which will be used to iterate through the images and compute the features for each fruit class separately
fnames = ['Apples','Oranges','Peaches']

i = 0 #Variable to keep track of the feature index for storing the computed features in the corresponding position in the feature vectors; will be incremented for each image processed. 
l = -1 
for img in imgs_filenames:
    l=l+1 # variable to keep track of the fruit class 
    d=0; # variable to keep track of the image number for the current fruit class 
    for path in img:
        labels.append(fnames[l])
        
        I = io.imread(path)
        #fig = px.imshow(I)
        #fig.show()

        # --------------------
        # Color Features
        # --------------------

        #Compute the color ratio feature - Ratio between average red and average green in a 100x100 region of interest (ROI) centered in the image (this ROI may need to be adjusted depending on the images)
        cr[i]=np.mean(I[150:250,150:250,0].flatten())/np.mean(I[150:250,150:250,1].flatten()) 

        # Compute mean hue, saturation and value for the whole image based on the HSV color space representation of the image
        hsv = cv.cvtColor(I, cv.COLOR_BGR2HSV) #Convert the image from BGR to HSV color space, which is often more suitable for color-based feature extraction.
        m_h[i]=np.mean(hsv[:, :, 0])
        m_s[i]=np.mean(hsv[:, :, 1])
        m_v[i]=np.mean(hsv[:, :, 2])

        # --------------------
        # Shape Features
        # --------------------

        #Binarize image for the shape ratio computation
        Igc = cv.cvtColor(I,cv.COLOR_BGR2GRAY) #Convert to grayscale 
        Ieq = cv.equalizeHist(Igc) # Equalize histogram to improve contrast and make thresholding more effective
        ret,Ith = cv.threshold(Ieq,100,255,cv.THRESH_BINARY) # Binarize image using a fixed threshold of 100 - this value may need to be adjusted depending on the images

        #Find the indexes of the contour
        [row,col] = np.where(np.diff(Ith[0:-5,0:-5])!=0) # slicing off the last 5 rows and columns before taking differences and avoid border artifacts produced by thresholding/equalization (shadows on the borders of the image that are not part of the fruit contour but may be detected as such due to the processing steps)
      
        # Compute the shape ratio
        sr[i]=(max(col)-min(col))/(max(row)-min(row)) 

        if d==5:#Plot just a single example image
            
            print("=================="+fnames[l]+"================")
            fig = px.imshow(Ith, title="Binarized Image")
            fig.show()
            
            fig = px.imshow(I, title="Original Image")
            
            #Show region used for color ratio
            fig.add_shape(type="line",x0=150, y0=150, x1=250, y1=150,
                          line=dict(color="Black",width=4,dash="dot",))
            fig.add_shape(type="line",x0=150, y0=250, x1=250, y1=250,
                          line=dict(color="Black",width=4,dash="dot",))
            fig.add_shape(type="line",x0=150, y0=150, x1=150, y1=250,
                          line=dict(color="Black",width=4,dash="dot",))
            fig.add_shape(type="line",x0=250, y0=150, x1=250, y1=250,
                          line=dict(color="Black",width=4,dash="dot",))
            fig.add_annotation(x=200,y=200,showarrow=False,text="Color Ratio ROI")

            #Plot height and width limits used for shape ratio
            fig.add_shape(type="line",x0=min(col), y0=min(row), x1=max(col), y1=min(row),
                          line=dict(color="LightSeaGreen",width=4,dash="dashdot",))
            fig.add_shape(type="line",x0=min(col), y0=max(row), x1=max(col), y1=max(row),
                          line=dict(color="LightSeaGreen",width=4,dash="dashdot",))
            fig.add_shape(type="line",x0=min(col), y0=min(row), x1=min(col), y1=max(row),
                          line=dict(color="LightSeaGreen",width=4,dash="dashdot",))
            fig.add_shape(type="line",x0=max(col), y0=min(row), x1=max(col), y1=max(row),
                          line=dict(color="LightSeaGreen",width=4,dash="dashdot",))
            fig.add_annotation(x=min(col)-10, y=150,showarrow=False,text="h")
            fig.add_annotation(x=150,y=max(row)+10,showarrow=False,text="w")
            fig.add_annotation(x=150,y=10,showarrow=False,text="ratio=w/h")

            fig.show()
            
            #Show computed contour
            fig=px.scatter(x=col,y=row,width=500, height=500, title="Contour Points")
            fig.show()
            
        i=i+1 #Update feature index
        d=d+1 #Update image index for the current fruit class
    

# Features plotting - represent fruits in the feature space

fig=px.scatter(x=cr,y=sr,color=labels,labels=dict(x="Color Ratio",
                                                  y="Shape Ratio",color="Fruit"), title="Fruits in the Feature Space")
fig.update_traces(marker_size=15)
fig.show()


fig2=px.scatter(x=cr,y=m_v,color=labels,labels=dict(x="Color Ratio",
                                                    y="Mean Value",color="Fruit"), title="Fruits in the Feature Space")
fig2.update_traces(marker_size=15)
fig2.show()
```
