```python
## FRUITS ##
from PIL import Image  # lib -> pip install pillow
import os

folder = ["FRUITS/FRUITS/APPLES/", "FRUITS/FRUITS/ORANGES/", "FRUITS/FRUITS/PEACHES/"]
labels = ["Apple", "Orange", "Peach"]
X = []  # features
y = []  # labels

for i in range(0,3):
    for file in [f for f in os.listdir(folder[i]) if f.endswith(".JPG")]:
        img = Image.open(os.path.join(folder[i], file))
        arr = np.array(img)

        R = arr[:, :, 0]
        G = arr[:, :, 1]
        B = arr[:, :, 2]

        features = [R.mean(), G.mean(), B.mean()]
        X.append(features)
        y.append(labels[i])  # A, O, P

dataFrame = pd.DataFrame(X, columns=["R_mean", "G_mean", "B_mean"])
dataFrame["label"] = y

dataFrame.to_csv("Fruits_featurs.csv")

px.scatter_3d(dataFrame, "R_mean", "G_mean", "B_mean", color="label").show()
```
